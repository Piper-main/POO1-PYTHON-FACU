Sistema de gestión de productos


1. Desarrollo de clases
Se desea desarrollar un sistema para gestionar productos de un comercio.

a) Crear una clase Producto que contenga los atributos nombre, codigo y precio_unitario, con valores predeterminados '', '' y 0, respectivamente. La clase debe contar con un método info() que permita mostrar la información del producto.

b) Crear una clase ProductoPerecedero que herede de Producto y agregue el atributo dias_vencimiento, con valor predeterminado 0. Debe implementar el método info() para mostrar toda la información correspondiente.

c) Crear una clase ProductoElectronico que herede de Producto y agregue el atributo meses_garantia, con valor predeterminado 0. Debe implementar el método info() para mostrar toda la información correspondiente.

d) Implementar en la clase Producto un método calcular_costo(cantidad) que devuelva el costo total correspondiente a la cantidad de unidades solicitadas.

e) Crear al menos dos objetos de cada tipo de producto y mostrar su información.


2. Función externa
Implementar una función llamada calcular_total_compra(lista_productos, cantidades) que reciba una lista de productos y una lista con las cantidades correspondientes a cada producto. La función deberá calcular y devolver el importe total de la compra.


3. Programa principal
Desarrollar un programa principal que permita probar las clases implementadas, mostrar la información de los productos y calcular el importe total de una compra.
