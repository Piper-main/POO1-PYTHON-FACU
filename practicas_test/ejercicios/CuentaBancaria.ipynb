1. Desarrollar un programa en Python que permita representar distintos tipos de cuentas bancarias.

- Crear una clase CuentaBancaria que no pueda ser instanciada, con los atributos titular, numero_cuenta y saldo (por defecto 0).
- Implementar el método info() para mostrar los datos de la cuenta.
- Implementar una propiedad saldo y su correspondiente setter. El saldo no puede ser negativo.
- Implementar el método depositar(monto), que permita incrementar el saldo de la cuenta.
- Crear una clase CuentaAhorro que herede de CuentaBancaria y agregue el atributo tasa_interes (por defecto 0).
- Crear una clase CuentaCorriente que herede de CuentaBancaria y agregue el atributo limite_descubierto (por defecto 0).
- Sobrescribir el método info() en ambas clases para mostrar todos los datos correspondientes.
- Crear dos objetos de la clase CuentaAhorro y dos de la clase CuentaCorriente, y mostrar su información.


2. Implementar una función calcular_saldo_total(lista_cuentas) que reciba una lista de cuentas bancarias y devuelva la suma de los saldos de todas las cuentas.


3. Crear un programa principal que permita probar el funcionamiento de las clases, las propiedades, los métodos y la función implementada.
